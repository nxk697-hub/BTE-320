```python
# Inputs: sales ($) and cost of goods pct (%)
sales = input("Enter sales ($): ")
sales = float(sales)

cogs_pct = input("Enter Cost of Goods (%): ")
cogs_pct = float(cogs_pct)

rate = input("Enter tax rate (%): ")
rate = float(rate)

# Processes: calculate cost of goods sold and gross profit
cogs = sales * (cogs_pct / 100)
grossProfit = sales - cogs
taxes = grossProfit * rate / 100
netIncome = grossProfit - taxes

# Outputs: display cost of goods sold and gross profit
print(f"Cost of Goods sold: ${cogs}")
print(f"Gross Profit: ${grossProfit}")
print(f"Taxes: ${taxes}")
print(f"Net income: ${netIncome}")
```

```python
def getValues(n: int) -> list:
  '''
  Creates a list of weights. The weights are given by the user
  Input:
  - number of item in the list
  Output:
  - a list of weights
  '''
  items = []
  for i in range(n):
    try:
      item = float(input('Enter weight value:'))
    except Exception as e:
      print(e)
      item = float('nan')

    items.append(item)

  return items

def calcAverage(alist: list) -> float:
  '''
  Input:
  - A list of weights
  Output:
  - Average value; excludes the first 2 items
  '''
  try:
    return sum(alist[2:])/len(alist[2:])
  except ZeroDivisionError:
    return float('nan')

def main():
  num = int(input('Enter number of weights:'))

  if num <= 0:
    raise ValueError('num must be positive')

  weights = getValues(num)
  avg = calcAverage(weights)
  print(f'Average weight: {avg} lbs.')

main()
```

```python
#Part A
def roi(cost:float, revenue:float)->float:
  assert cost > 0, "the cost must be positive!"
  assert isinstance(cost, (int,float)), "cost must ne numerical"
  assert isinstance(revenue, (int,float)), "revenue must be numerical"

  return (revenue - cost) / cost

if __name__ =="__main__":
  c=float(input('Enter cost value:'))
  r=float(input('Enter revenue value:'))
  retInvest=roi(c,r)
  print(retInvest)
```

```python
class Greeting:
  def __init__(self,first, last):
    self.first=first
    self.last=last

  def __str__(self):
    return "Instance of class Greeting"

  def display(self, reverse=True):
    if reverse:
      print(f'{self.last}, {self.first}')
    else:
      print(self.first)

name = Greeting('John', 'Doe')
print(name)
name.display()
name.display(reverse=False)

```

```python
class Student:
  def __init__(self, name, number):
    self.__name = name
    self.__number = number
    self.__courses = []

  def get_name(self):
    return self.__name

  def get_courses(self):
    return self.__courses

  def set_name(self, newname):
    self.__name = newname

  def __len__(self):
    return len(self.__courses)

  def enroll(self, course):
    if course in self.__courses:
      print('You have already enrolled in', course)
    else:
      self.__courses.append(course)

s = Student('Bob', 12345)
s.get_name()
s.get_courses()
s.set_name('Bob Andrews')
s.get_name()
s.enroll('ABC123')
s.enroll('ABC123')
s.enroll('ABC567')
len(s)
s.get_courses()
```

```python
class Point:
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def distance(self, other):
    return ((self.x - other.x) ** 2 + (self.y - other.y) ** 2) ** .5

class Rectangle:
  def __init__(self):
    self.p1 = Point(0,0)
    self.p2 = Point(4,0)
    self.p3 = Point(4,4)
    self.p4 = Point(0,4)
    self.width = self.p1.distance(self.p2)
    self.height = self.p2.distance(self.p3)

  def area(self):
    return self.width * self.height

  print(Rectangle().area())
```

```python
def loan_payment(principal: float, annual_rate: float, months: int) ->float:
  assert isinstance(principal, (int, float)), "principal must be numeric"
  assert isinstance(annual_rate, (int, float)), "annual_rate must be numeric"
  assert isinstance(months,int), "months must be an integer"

  if principal <= 0:
    raise ValueError("Principal must be > 0.")
  if annual_rate < 0:
    raise ValueError("Annual rate cannot be negative")
  if months <= 0:
    raise ValueError("Number of months must be > 0.")

  r = annual_rate / 12.0
  n=months
  P=float(principal)

  if r==0.0:
    return P / n
  else:
    return r * P / (1.0 -(1.0 + r) ** (-n))

if __name__== "__main__":
  try:
    P = float(input("Enter Principal: "))
    apr = float(input("Enter annual interest rate as a decimal: "))
    n = int(input("Enter number of months: "))
    payment = loan_payment(P, apr, n)
    print(f"Monthly payment: {payment:.2f}")
  except Exception as e:
    print(e)
```
